### End-to-end drug discovery pipeline

From data collection to optimization, this notebook demonstrates the
complete workflow used in real drug discovery campaigns:

1. Data Collection: fetch ChEMBL bioactivities for target
2. Featurization: convert SMILES to numerical representations
3. Model Training: train ensemble with Optuna optimization
4. Explainability: SHAP analysis of predictions
5. Virtual Screening: filter ZINC library for actives
6. Chemical Space: visualize UMAP and similarity search
7. Optimization: hill-climbing to improve predicted affinity
