### What is ChEMBL and where does this data come from

ChEMBL is a database of bioactive molecules with drug-like properties,
maintained by EMBL-EBI. It contains bioactivity data from scientific
literature, covering millions of compounds and thousands of targets.
The data comes from published assays measuring binding affinity, inhibition,
and other pharmacological properties.

### Activity distribution per target

# Load all_targets.csv and show shapes
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('data/raw/all_targets.csv')
print(f"Total compounds: {len(df)}")
print(f"Targets: {df['target_name'].unique()}")

# Box plots of pChEMBL per target
plt.figure(figsize=(12, 6))
sns.boxplot(x='target_name', y='pchembl_value', data=df)
plt.xticks(rotation=45)
plt.title('pChEMBL Distribution by Target')
plt.tight_layout()
plt.show()

### Dataset overlap: compounds in multiple targets
